```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
data = {
    'Age': [18, 20, 22, 25, 27, 30, 45, 48, 50, 52],
    'Income': [20, 22, 25, 28, 30, 35, 70, 75, 80, 85]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.head()
```

```python
x=df[["Age","Income"]]
```

```python
#scaling
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
x_scaled
```

```python
#model
from sklearn.cluster import KMeans
```

```python
model=KMeans(
    n_clusters=3,random_state=42
)
model.fit(x_scaled)
```

```python
df['Cluster']=model.labels_
print(df)
```

```python
plt.scatter(
    x_scaled[:, 0],
    x_scaled[:, 1],
    c=model.labels_
)

plt.xlabel('Age')
plt.ylabel('Income')
plt.title('K-Means Clustering')

plt.show()
```
