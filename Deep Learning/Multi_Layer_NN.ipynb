```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# 1 ---> Datasets
x=torch.tensor([[1.0],[2.0],[3.0],[4.0]])
y=torch.tensor([[2.0],[4.0],[6.0],[8.0]])

# 2 ---> Multilayer NN
model = nn.Sequential(
    nn.Linear(1,4),
    nn.LeakyReLU(),

    nn.Linear(4,4),
    nn.LeakyReLU(),

    nn.Linear(4,1)
)
# 3 -->Loss Function
loss_function = nn.MSELoss()

# 4 ---> optimizer
optimizer = optim.SGD(
    model.parameters(),
    lr=0.01
)
# 5 -->train the model
for epoch in range(500):
    #Forward pass
    y_pred = model(x)

    #loss
    loss = loss_function(y_pred,y)

    #clear old gradient
    optimizer.zero_grad()

    #backward
    loss.backward()

    #update weights
    optimizer.step()

    #print
    if (epoch % 50 == 0):
        print(
            f"Epoch {epoch + 1} --> "
            f"Loss : {loss.item():.4f}"
        )

# 6 -->Prediction
with torch.no_grad():
    test=torch.tensor([[5.0]])
    prediction=model(test)
    print("Prediction : ",prediction.item())
```
