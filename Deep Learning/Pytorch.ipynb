Gradient batata hai ki kisi value ko thoda change karne par output kitna aur kis direction mein change hoga.

PyTorch Autograd --> Automatically calculate gradient

```python
import torch
x = torch.tensor(2.0, requires_grad=True)  #It tell us that --- PyTorch, x ke saath hone wale calculations ko track karo, kyunki mujhe iska gradient calculate karna hai
y = x ** 2
y.backward()
print(x.grad)
```

```python
weight = torch.tensor(2.0, requires_grad=True)
prediction = weight * 3
loss = (prediction - 10) ** 2
loss.backward()
print(weight.grad)
```
