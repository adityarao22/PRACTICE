Gradient batata hai ki kisi value ko thoda change karne par output kitna aur kis direction mein change hoga.

PyTorch Autograd --> Automatically calculate gradient

```python
import torch
x = torch.tensor(2.0, requires_grad=True)  #It tell us that --- PyTorch, x ke saath hone wale calculations ko track karo, kyunki mujhe iska gradient calculate karna hai
y = x ** 2
y.backward()
print(x.grad)
```

```python
weight = torch.tensor(2.0, requires_grad=True)
prediction = weight * 3
loss = (prediction - 10) ** 2
loss.backward()
print(weight.grad)
```

```python
import torch
# Weight
w = torch.tensor(5.0, requires_grad=True)
# Learning rate
learning_rate = 0.1
# Input
x = torch.tensor(2.0)
# Actual output
y_true = torch.tensor(4.0)
# Training
for epoch in range(10):
    # Forward pass
    y_pred = w * x
    # Loss
    loss = (y_pred - y_true) ** 2
    # Backpropagation
    loss.backward()
    # Weight update
    with torch.no_grad():
        w -= learning_rate * w.grad
    # Gradient reset
    w.grad.zero_()
    print(
        "Epoch:", epoch + 1,
        "Loss:", loss.item(),
        "Weight:", w.item()
    )
```

Neural Network   ----> Single Neuron Network

```python
import torch 
import torch.nn as nn
import torch.optim as optim
```

```python
# 1 ---> Datasets
x=torch.tensor([[1.0],[2.0],[3.0],[4.0]])
y=torch.tensor([[2.0],[4.0],[6.0],[8.0]])

# 2 --> Create model
model = nn.Linear(1,1)

# 3 ---> Loss Function
loss_function = nn.MSELoss()

# 4 ---> Optimizer
optimizer = optim.SGD(
    model.parameters(),
    lr=0.01
)

# 5 ---> Training loop
for epoch in range(100):
    # Forward pass
    y_pred=model(x)

    #Loss
    loss=loss_function(y_pred,y)

    #remove old gradient
    optimizer.zero_grad()

    # Backward propagation
    loss.backward()

    # Update weight and bias
    optimizer.step()

    #print loss in every  10 step
    if (epoch + 1) % 10 == 0:
        print(
            f"epoch{epoch + 1} , "
            f"Loss : {loss.item():.4f}"
        )

# 6 ---> Prediction
with torch.no_grad():
    test = torch.tensor([[5.0]])
    prediction=model(test)
    print("Prediction for x=5.0 :",prediction.item())
    
```
