```python
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("random_forest_placement_dataset.csv")
df
```

```python
#feature selection
x=df.drop("Placement",axis=1)
y=df["Placement"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from xgboost import XGBClassifier
```

```python
model=XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)
```

```python
#train model
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
print("Actual value:",y_test)
print("Predicted value:",y_pred)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
print("Accuracy score:\n",accuracy_score(y_test,y_pred))
print("Confusion matrix:\n",confusion_matrix(y_test,y_pred))
print("classification_report:\n",classification_report(y_test,y_pred))
```

Hyperparameter

```python
from sklearn.model_selection import GridSearchCV
from xgboost import XGBClassifier
```

```python
model=XGBClassifier(random_state=42)
```

```python
#hyperparameter
param_grid={
    "n_estimator":[50,100,200],
    "learning_rate":[0.01,0.1,0.2],
    "max_depth":[3,5,7]
}
```

```python
grid_search=GridSearchCV(
    model,
    param_grid,
    cv=5,
    scoring="accuracy"
)
```

```python
grid_search.fit(x_train,y_train)
```

```python
#best parameter
print("Best Parameter:",grid_search.best_params_)
```

```python
#best score
print("Best Score:",grid_search.best_score_)
```

```python
#best model
best_model=grid_search.best_estimator_
```

```python
print("Best model:",best_model.fit(x_train,y_train))
```

```python
best_model
```

```python
#prediction
y_pred_best_model=best_model.predict(x_test)
```

```python
print("Prediction of best model:",y_pred_best_model)
```

```python
#final Accuracy
print("Accuracy of Final Model:",accuracy_score(y_test,y_pred_best_model))
```

Accuracy Of RandomForest VS XgBoost

```python
rf_accuracy=accuracy_score(y_test,y_pred)
xgb_accuracy=accuracy_score(y_test,y_pred_best_model)
print("Random Forest Accuracy:",rf_accuracy*100,"%")
print("XgBoost Accuracy:",xgb_accuracy*100,"%")
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("Random Forest")
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))

print("XGBoost")
print(confusion_matrix(y_test, y_pred_best_model))
print(classification_report(y_test, y_pred_best_model))
```
