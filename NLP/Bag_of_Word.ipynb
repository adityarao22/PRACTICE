```python
import nltk
import string
```

```python
nltk.download("punkt_tab")
```

```python
documents = [
    "I love learning Machine Learning!",
    "I am studying NLP and Machine Learning.",
    "NLP is interesting and powerful!"
]
```

```python
from nltk import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer
```

```python
nltk.download("stopwords")
```

```python
stop_words=set(stopwords.words("english"))
```

```python
#preprocess ---> it convert sentence into text
from string import punctuation


def preprocess(text):
    #tokenization
    tokens=word_tokenize(text)

    #Lower_Case + stopwords  + remove punctuation
    clean_tokens=[word.lower() for word in tokens if word.lower() not in stop_words and word not in string.punctuation]

    # convert token into sentence
    return " ".join(clean_tokens)

#preprocess all documents
clean_documents=[
    preprocess(doc)
    for doc in documents
]

print("clean_documents")
print(clean_documents)

```

```python
#Bag Of Words
vectorizor=CountVectorizer(ngram_range=(1,2))
```

```python
x=vectorizor.fit_transform(clean_documents)
```

```python
print("\nVocabulary")
print(vectorizor.get_feature_names_out())
```

```python
#Bow Matrix
print(x.toarray())
```
