```python
import nltk
nltk.download("punkt_tab")
from nltk.tokenize import word_tokenize

text="I am learning NLP, and I am studying Machine Learning!"

tokens=word_tokenize(text)
print(tokens)
```

```python
tokens_lower=[word.lower() for word in tokens]

print(tokens_lower)
```

```python
nltk.download("stopwords")

from nltk.corpus import stopwords

stop_words=set(stopwords.words("english"))
filter_tokens=[word for word in tokens_lower if word not in stop_words]
print(filter_tokens)
```

```python
import string
```

```python
#remove punctuation
filter_token=[word for word in tokens_lower if word not in stop_words and word not in string.punctuation]

print(filter_token)
```

```python
#stemming
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

stemmed_words=[stemmer.stem(word) for word in filter_token]

print(stemmed_words)
```

```python
#Lemmatization
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
from nltk.stem import WordNetLemmatizer
Lemmatize=WordNetLemmatizer()

words=['learning', 'nlp', 'studying', 'machine', 'learning']
Lemmatized_word=[Lemmatize.lemmatize(word,pos="v") for word in words]
print(Lemmatized_word)
```
