```python
from gensim.models import Word2Vec
```

```python
sentences = [
    ["i", "love", "machine", "learning"],
    ["i", "love", "deep", "learning"],
    ["machine", "learning", "is", "powerful"],
    ["deep", "learning", "is", "interesting"]
]
```

```python
model=Word2Vec(
    sentences,
    vector_size=10, #size 10 Dimensional ka vector bnega hr word ka
    window=2,      #HR WORD KO 2-2 CONTEXT ME DEKHEGA
    min_count=1, #word agr ek baar v aaya ho to usko vocab me add kro
    sg=0    # sg=0 -->CBOW  sg=1 -->Skip-Gram
)
```

```python
print("/nVocabulary")
print(model.wv.index_to_key)
```

```python
print("/Vector for machine")
print(model.wv["machine"])
```

```python
print("Similar word")
print(model.wv.most_similar("machine"))
```
