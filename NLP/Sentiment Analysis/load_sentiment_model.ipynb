```python
#load the model
import joblib
model=joblib.load("sentiment_model.pkl")
vectorizor=joblib.load("Tf-idf_vectorizer.pkl")
```

```python
new_text=["The product quality is amazing and I love it"]
new_text_vc=vectorizor.transform(new_text)
new_pred=model.predict(new_text_vc)
print("Prediction :",new_pred)
```
